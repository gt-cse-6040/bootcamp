## Debugging your functions with code

### Lists/Sets

```python
def lists_sets_test(key_name):
    # assign the test case variables
    my_out = returned_output_vars[key_name]
    true_out = true_output_vars[key_name]

    # Are the lengths the same?
    display(len(my_out), len(true_out))
    
    # Are the values the same?
    display(my_out == true_out)
    
    # Items in yours but not expected
    # Returns the difference between the two
    display(set(my_out) - set(true_out))

    return True
    
```

### Dictionaries

```python
def dicts_test(key_name):
    # assign the test case variables
    my_out = returned_output_vars[key_name]
    true_out = true_output_vars[key_name]
    
    # gives true/false result if keys are same or different
    display(my_out.keys() == true_out.keys())
    
    # Tells which keys are different
    display([k for k in my_out if my_out[k] != true_out[k]])  # differing keys
    
    
    # tells which keys have differences in their values
    for key in true_out:
            if true_out[key] != my_out[key]:
                print('returned_output_vars values not equal to true_output_vars values for the key: ',  true_out[key])
                display('true_output_vars key')
                display(true_out[key])
                display('returned_output_vars key')
                display(my_out[key])
                break

    return True
```

### Numbers/Floats

```python
def numbers_test(key_name):
    # assign the test case variables
    my_out = returned_output_vars[key_name]
    true_out = true_output_vars[key_name]
    
    display(abs(my_out - true_out) < 1e-6)
    return True
```

### Strings

```python
def string_test(key_name):

    from itertools import zip_longest
    
    # assign the test case variables
    my_out = returned_output_vars[key_name]
    true_out = true_output_vars[key_name]

    display(my_out)
    display(true_out)
    display([int(c0==c1) for c0, c1 in zip_longest(my_out, true_out)])
    

    return True
```
