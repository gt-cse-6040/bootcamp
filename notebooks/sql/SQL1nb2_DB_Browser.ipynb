# Query Editors 

Rather than using Vocareum or Google Collab and only seeing a few rows and columns, you can use query editors to work on queries. If you're familiar with Microsoft SQL, their query editor is SQL Server Management Studio. Oracle uses Oracle SQL Developer or SAP uses HANA Studio. For SQLite databases, we can use [DB Browser for SQLite](https://sqlitebrowser.org/).

After installing DB Browser, you can either download any sqlite database that peaks your interest, or you can click on the database link in the markdown box above this to download it.

Once you download it, open DB Browser and click on File > Open database > (select database you downloaded)

On the first tab, Database Structure, it shows a list of all the tables, and the CREATE TABLE statements that would be used to generate the tables in question.



On the second tab, Browse Data, you can click on Table and select a table from the database. It shows all of the rows / columns of the table you select. It's functionally the same as writing 'select * from graduation'

Edit Pragmas is a tab you don't need to worry about. It gives you options to change settings for the query engine.



Excute SQL is the tab where the fun stuff happens! From here you can write queries to query against the database you have loaded. It has autocomplete for object names.

To execute your query there's a few ways to do it. Press F5 on your keyboard or press the play button icon in the top center.
If you have multiple queries in the same window, or want to run just a subquery or CTE, you can highlight the code you want executed with your mouse / keyboard and it will only execute what you've highlighted.

You can open as many query windows as your computer can handle. You can do that with CTRL-T(Windows PCs) or pressing the new query button above SQL 1.

Once you've executed your query, the yellow circle is your query results.
The green circle shows any errors in your code as applicable and the execution results if succcessful.
